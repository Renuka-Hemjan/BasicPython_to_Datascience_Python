```python
html = """
<html>
<head>
    <title>Learn With Rakesh</title>
</head>
<body>
    <h1>Welcome to Web Scraping</h1>
    <p class="info">Learning BeautifulSoup</p>
    <h3>this is h3 tag in body section </h3>

</body>
<footer> <h3>copyright @ rakesh.com </h3> </footer>
</html>
"""
```

```python
from bs4 import BeautifulSoup #BeautifulSoup takes raw HTML and converts it into a structured object 
#that Python can easily search and navigate.
```

```python
soup = BeautifulSoup(html, 'html.parser')# html.parser read the HTML code and identify its structure and tags.
print(soup.title.text)
```

```python
print(soup.title)
```

```python
print(soup.h1.text)
```

```python
print(soup.h3.text)
```

```python
print(soup.footer.h3.text)
```

```python
allh3 = soup.find_all("h3")
for t in allh3:
    print(t.text)
```

```python
import requests#To send requests to websites and get a response from them.
```

```python
url = 'http://techaxis.com.np/'
response = requests.get(url)
response
```

```python
soup = BeautifulSoup(response.text,"html.parser")
print(soup.title.text)
```

```python
# print(soup.body.text)
```

```python
h1 = soup.find('h1')
if h1:
    print(f' h1 tag found:{h1.text}')
else:
    print('No h1 tag found')
```

```python
h3 = soup.find('h3')
if h3:
    print(f' h3 tag found:{h3.text}')
else:
    print('No h3 tag found')
```

```python
h2 = soup.find('h2')
if h2:
    print(f' h2 tag found:{h2.text}')
else:
    print('No h2 tag found')
```

```python
h2_tags = soup.find_all("h3")
for i, tag in enumerate(h2_tags,start=1):
    print(f"{i}. {tag.get_text(strip=True)}")
    
print(len(h2_tags))    
```

```python
url = "https://techaxis.com.np/"
res = requests.get(url)

soup = BeautifulSoup(res.text, "html.parser")

tags = soup.find_all(["h1","h2","h3"])

for tag in tags:
    print(tag.name, ":",tag.get_text(strip=True))
```

```python
allh= soup.find_all(["h1","h2","h3",'p'])
for p in allh:
    print(p.text)
print(allh)    
```

```python
allh= soup.find_all(["p"])
for p in allh:
    print('p: ',p.text)  
```
