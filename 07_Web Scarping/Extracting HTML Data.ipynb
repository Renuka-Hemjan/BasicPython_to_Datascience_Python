```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.scrapethissite.com/pages/"
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
headings = soup.find_all("h3")
```

```python
for heading in headings:
    #print("Title:", heading.get_text(strip=True))
    print("\033[1mTitle:", heading.get_text(strip=True),"\033[0m")
    description = heading.find_next("p")
    if description:
        print("Description:", description.get_text(strip=True))
        print()
```

```python
for tag in soup.find_all(["h1","h2","h3","p"]):
    if tag.name != 'p':
        print(tag.name, ":", tag.get_text(strip=True))
        
```

```python
for tag in soup.find_all(["h1","h2","h3","p"]):
    if tag.name != 'p':
        print(tag.name, ":", tag.get_text(strip=True))
    if tag.name == 'p':
        print(f'\t ---:{tag.get_text(strip=True)}')
        
```

```python
print(soup.find("div", class_="container").prettify())
```

```python
container = soup.find("div", class_="container")
print(container.get_text(" ", strip=True))
```

```python
container = soup.find("div", class_="container")
for c in container:
    print(container.get_text(" ", strip=True))
 
```

```python
for i, section in enumerate(soup.body.find_all(recursive=False), start=1):
    print(f"\n{'='*50}")
    print(f"Section {i}: <{section.name}>")
    print(f"{'='*50}")
    print(section.prettify())
```
