```python
#webscraping
introduction to web scraping and its ethics::

web scraping is th eprocess of automatically extracting data from websites using software or scripts instead
of manually copying information.
    
process of web scraping
-send a request to a websites
-downloads the webpages's html content.
-extract the required information
-stores the data in a structured format such as CSV,Excel, or a Database.
```

```python
Example
suppose you want to collect product names and prices from e-commerce website.
    instead of manually copying hundreds of prof=ducts, a scraper can automatically extract::
        product       price
        laptop a       500$
        laptop B       650$
        laptop c       700$
```

```python
#common python libraraies for web scraping
requests - download webpages content.
Beautifulsoup - parses HTML and extracts data.
Scrapy - powerful web scraping framework.
selenium - Automates browsers and handles javaScript- heavy websites.
pandas - stores and analyzes scraped data
```

```python
Ethics of web scraping::
    just beacause data is visible on a website doesnot alaways mean it should be scraped without comsideration.
1. Respect Website terms of service
Many websites specify whether scraping is allowed in their terms of service
Read  the websites's policies.
follow any restrictitons they provide.
```

```python
2. check robots.txt
Many websites publish a robot.txt file indicating which areas automated systems should avoid.
(robots.txt is a plain text file placed at the root of a website(for example, 
https://rakesh.com/robots.txt)tells web crawlers and search engines bots which parts of the site they are allowed or not allowed
to access.)
Example:
```

```python
User-agent = *
Disallow: /admin/
Disallow: /private/
Allow: /

Sitemap: https://example.com/sitemap.xml
----------------------------------------
https://rakesh.com/robots.txt
A robots.txt file may contain:
User-agent:*
Disallow: /private/
```

```python
Common directives::
User-agent-specifices which crawler the rules apply to (*means all crawlers).
Disallow - tells crawlers not to visit a path.
Allow -  explicity permits access to a path
Sitemap - points crawlers to the site's map.
```

```python
4.Respect Privacy: Do not collect " Personal information, Passwords,Private user data,
Sensitive information
unless you have proper authorization and a legitimate reason.
```

```python
5. Use Public ApIs when available:
Many websites provide APIs specifically for accessing data.
Benefits: More reliable,Faster,Usaually allowed by the provider,Strucutured data format(JSON)
```

```python
6. Follow Copyright and Data Usage Rules::
Even if you can scrape content, you may not have the right to republish or redistribute it.
Always check: Copyright policies, Licensing terms, Data usage restrictions.
```
