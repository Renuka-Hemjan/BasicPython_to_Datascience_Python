```python
# time series analysis is the process of analyzing data collected over time to 
# identify patterns, trends, seasonal effects nad make future predictions.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings("ignore")
```

```python
data = {
    "Date": pd.date_range(start="2024-01-01", periods=24, freq="ME"),
    "Sales": [ 120,130,128,135,145,150,
               160,170,175,180,190,200,
               205,210,220,225,230,240,
               250,255,260,270,275,280]
}
df = pd.DataFrame(data)
df.set_index('Date', inplace=True)
print(df.head())
```

```python
data = {
    "Date": pd.date_range(start="2024-01-01", periods=24, freq="MS"),#month start
    "Sales": [ 120,130,128,135,145,150,
               160,170,175,180,190,200,
               205,210,220,225,230,240,
               250,255,260,270,275,280]
}
df = pd.DataFrame(data)
df.set_index('Date', inplace=True)
print(df.head())
```

```python
plt.figure(figsize=(10,5))
plt.plot(df.index,
         df['Sales'],
         marker='o',
         linestyle='-',
         linewidth=2)

plt.title("Monthly Sales Time Series")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df.index,
         df['Sales'],
         marker='o',
         linestyle='-',
         linewidth=2)

plt.title("Monthly Sales Time Series")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
```

```python
#monthly change
df["Sales_Change"] = df["Sales"].diff()
print(df)
```

```python
df["Growth%"] = df["Sales"].pct_change() * 100
print(df)
```

```python
df["Moving_Average_3M"] = df["Sales"].rolling(window=3).mean()
print(df)
```
