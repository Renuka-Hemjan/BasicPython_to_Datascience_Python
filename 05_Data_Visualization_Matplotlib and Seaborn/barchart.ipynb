```python
import matplotlib.pyplot as plt
courses = ["Python", "Java", "C", "Datascience"]
students_count = [40, 30, 20, 50]
plt.figure(figsize=(6,4))
plt.bar(courses, students_count)
plt.title('Courses Vs Study_hours')
plt.xlabel('Name of Courses')
plt.ylabel('Student Count')
plt.show()
```

```python
import pandas as pd
df = pd.DataFrame({
    
    "Department":["IT","HR","Sales","Finance"],
     "Employees":[25,15,30,20]
                  })
plt.figure(figsize=(6,4))
plt.barh(df['Department'],df['Employees'])# barh gives bar left to right
plt.title('Department Vs employees')
plt.xlabel('employees')
plt.ylabel('department')
plt.savefig('barcht.png')#this code saves the bar in imange in folder or file
plt.show()
        
```

```python
import numpy as np
subjects = ["DSA","OOP","Database"]
#value database
boys = [80,75,90]
girls =[85,80,88]
x = np.arange(len(subjects))#[0 1 2 3]
width = 0.40
plt.figure(figsize=(6,6))
plt.bar(x - width/2,boys,width,label="Boys")
plt.bar(x + width/2,girls,width,label="Girls")
plt.title('Marks by subject')
plt.xlabel('subjects')
plt.ylabel('marks')

plt.xticks(x,subjects)
plt.legend()
plt.grid(axis="y")
plt.show()
```
