```python
import seaborn as sns
import pandas as pd
```

```python
tips= sns.load_dataset('tips')
df = tips.head()
df
```

```python
corr =tips.corr(numeric_only=True)# value near to 1 play vital role in ml
print(corr) 
```

```python
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
corr = tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True)
plt.show()
```

```python
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
corr = tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm')
plt.show()
```

```python
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
corr = tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm', fmt='.2f')
plt.show()
```

```python
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
corr = tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm', fmt='.2f' , linewidth=1)
plt.show()
```

```python
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
corr = tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm', fmt='.2f' , linewidth=1,alpha=0.7) #alpha gives opacity
plt.title('tips correlation')
plt.show()
```

```python
df = pd.DataFrame({
    "Age":[22,25,30,35,40],
    "Salary":[25000,30000,45000,55000,70000],
    "Experience":[1,3,5,8,12]
})
```

```python
corr =df.corr(numeric_only=True)
print(corr)
```

```python
df
corr = df.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm', fmt='.2f' , linewidth=1,alpha=1) #alpha gives opacity
plt.title(' age,experience and salary correlation')
plt.show()
```

```python
df = pd.DataFrame({
    "Experience":[1,2,3,4,5,6,7,8],
    "Salary":[25000,30000,35000,40000,45000,50000,55000,60000],
    "Stress":[80,75,70,65,60,55,50,45],
    "Luckey_No":[7,3,9,1,6,4,8,2]
})
print(df)
```

```python
corr =df.corr(numeric_only=True)
print(corr)
```

```python
df
corr = df.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr,annot = True, cmap = 'coolwarm', fmt='.2f' , linewidth=1,alpha=1) #alpha gives opacity
plt.title('positive , Negative and zero Correlation')
plt.show()
```
