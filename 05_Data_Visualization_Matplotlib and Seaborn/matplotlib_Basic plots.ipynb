```python
import matplotlib.pyplot as plt
```

```python
revenue = [45000,47000,55000,50000,46000]
year = [2015,2016,2017,2018,2019]
plt.plot(year,revenue)
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue)
plt.plot(year,expenses)
plt.style.use('default')#plt.style.use() = choose the look of your graph
# 'default' = normal look
plt.xticks(year)#remove decimal in year
plt.show()
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue, color = 'red')
plt.plot(year,expenses, color = 'green')
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expense')
plt.style.use('default')
plt.xticks(year)#remove decimal in year
plt.show()
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue, color = 'red', linestyle='dashed',linewidth=3)
plt.plot(year,expenses, color = 'green',linestyle='dotted',linewidth=3)
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expense')
plt.style.use('default')
plt.xticks(year)#remove decimal in year
plt.show()
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue, color = 'red', linestyle='dashed',linewidth=3, marker='D',markersize = '10')
plt.plot(year,expenses, color = 'green',linestyle='dotted',linewidth=3, marker='D')
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expense')
plt.style.use('default')
plt.xticks(year)#remove decimal in year
plt.show()
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue, color = 'red', linestyle='dashed',linewidth=3, marker='D',markersize = '10', label='revenue')
plt.plot(year,expenses, color = 'green',linestyle='dotted',linewidth=3, marker='D',label='expense')
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expense')
plt.style.use('default')
plt.xticks(year)#remove decimal in year
plt.legend(loc='best')
plt.show()
```

```python
revenue = [45000,47000,55000,50000,46000]
expenses = [20000,25000,35000, 40000,42000]
year = [2015,2016,2017,2018,2019]
plt.plot( year,revenue, color = 'red', linestyle='dashed',linewidth=3, marker='D',markersize = '10', label='revenue')
plt.plot(year,expenses, color = 'green',linestyle='dotted',linewidth=3, marker='D',label='expense')
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expense')
plt.style.use('default')
plt.xticks(year)#remove decimal in year
plt.legend(loc='best')#legend = tells us what each line/bar represents
# loc='best' = Matplotlib chooses the best place for the legend
plt.grid(True, linestyle='--')
plt.show()
```
