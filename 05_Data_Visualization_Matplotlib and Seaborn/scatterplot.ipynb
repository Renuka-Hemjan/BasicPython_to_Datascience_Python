```python
import matplotlib.pyplot as plt

#sample data
study_hours = [1,2,3,4,5,6,7]
exam_scores = [55,60,65,72,78,85,90]
plt.figure(figsize=(6,4))
plt.scatter(study_hours, exam_scores)
plt.show()
```

```python
study_hours = [1,2,3,4,5,6,7]
exam_scores = [55,60,65,72,78,85,90]
plt.figure(figsize=(6,4))
plt.scatter(study_hours, exam_scores)
plt.title('exam vs marks')
plt.xlabel('study hour')
plt.ylabel('marks')
plt.show()
```

```python
study_hours = [1,2,3,4,5,6,7]
exam_scores = [55,60,65,72,78,85,90]
plt.figure(figsize=(6,4))
plt.scatter(study_hours, exam_scores)
plt.title('exam vs marks')
plt.xlabel('study hour')
plt.ylabel('marks')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
df = pd.DataFrame({
    'StudyHours':study_hours,
    'Scores': exam_scores
})
df
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['StudyHours'], df['Scores'])
plt.title('exam vs marks')
plt.xlabel('study hour')
plt.ylabel('marks')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
df= sns.load_dataset('tips')
df
```

```python
plt.scatter(df['total_bill'],df['tip'])
plt.grid(True)
plt.title('total bills and tip relation')
plt.xlabel(' total bills')
plt.ylabel('tips')
plt.show()
            
```

```python
tips= sns.load_dataset('tips')
sns.scatterplot(data=tips , x='sex',y='tip', hue='sex')#different color
plt.show()
```
