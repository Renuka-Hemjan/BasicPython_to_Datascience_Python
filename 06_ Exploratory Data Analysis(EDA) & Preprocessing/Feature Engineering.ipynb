```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Age':[25,np.nan,30],
    'Salary':[50000,60000,70000]
})
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

```python
df = pd.DataFrame({
    'Age':[2,np.nan,30,28,35,np.nan],
    'Salary':[50000,60000,70000,55000,900000,65000],
    'Gender':['Male','Female','Male','Female','Female','Male'],
    'Department':[ 'HR','IT','Finance','IT','HR','Marketing'],
    'City':['Kathmandu','Pokhara','Kathmandu','Pokhara','Lalitpur','Birgunj']
})
df
```

```python
# create new feature#
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0,18,30,50,100],
    labels=['Young','Adult','Middle_Age','Senior']
)
print(df[['Age','Age_Group']])
```

```python
df['Salary_Level'] = pd.cut(
    df['Salary'],
    bins=[0,60000,100000,float('inf')],
    labels=['Low','Medium','High']
)
print(df[['Salary','Salary_Level']])
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(drop='first',sparse_output=False)
encoded = encoder.fit_transform(df[['Gender','Department','City']])
encoded
```

```python
df_encoded =pd.get_dummies(
    df,
    columns=['Gender','Department','City','Age_Group','Salary_Level'],
    dtype=int
)
print(df_encoded)
```

```python
encoded_df = pd.DataFrame(encoded,
                          columns=encoder.get_feature_names_out(['Gender','Department','City']))
encoded_df
```
