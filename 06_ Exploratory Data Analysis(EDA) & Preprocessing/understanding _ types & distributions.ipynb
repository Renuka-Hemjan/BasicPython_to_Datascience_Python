```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Height":[170.5,168.3,180.9,175.1,169.8],
    "Weight": [65.2,70.1,82.5,80.40,85.0],
    "Temperature":[36.8,37.2,36.5,40.0,42.0]
})
print(df.dtypes)
```

```python
plt.scatter(df["Height"],df["Weight"])
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

```python
# discrete data

df = pd.DataFrame({
    "Department":[
        "IT","HR","IT","Sales","HR",
        "IT","Sales","HR","IT","Sales"
    ]
})
```

```python
sns.countplot(data=df, x="Department")
plt.title("Employee count by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
import pandas as pd
df = pd.DataFrame({
    "Department":["IT","HR","Sales","Finance"],
    "Employees":[45,20,35,25]
})
print(df)
```

```python
sns.barplot(data=df, x="Department", y ="Employees")
plt.title("Barchart Employees count by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
            #histogram shows the history of numerical data by grouping them into groups called bins
```

```python
#nominal data
departments=["IT","HR","Sales","Finance"]
employees=[40,20,25,15]

plt.figure(figsize=(6,6))
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Employee Distribution by Department")
plt.axis("equal")
plt.show()
```
