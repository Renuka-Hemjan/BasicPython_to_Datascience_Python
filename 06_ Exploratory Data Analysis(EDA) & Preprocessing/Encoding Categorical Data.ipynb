```python
#encoding changing text data into numerical data.
# encoding catrgorial data: one hot ending and label encoding::
# categorial data is data that contain categories (text value)instead of number.
# types:
# one hot encoding (for nominal data) eg name of province
#One-hot encoding converts categories into separate columns containing 0 and 1.
# ordinal encoding(for ordinal)#in ml 3 bits data can be cahnge to 2
#Ordinal encoding converts categories that have a natural order/ranking into numbers.
```

```python
import pandas as pd
df = pd.DataFrame({
    "Color":["Red","Blue","Green","Blue","Red"]
})
print(df)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder()
encoded = encoder.fit_transform(df[['Color']])#The outer [] select columns, and the inner ['Color'] specifies the column name.
encoded
 #OneHotEncoder needs 2D because it expects data in the form of a table                               
```

```python
encoder = OneHotEncoder(sparse_output=False)#"Show me the encoded 0s and 1s normally."
encoded = encoder.fit_transform(df[['Color']])
encoded
                                
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['Color'])
)
print(encoded_df)
```

```python
encoder = OneHotEncoder(drop='first', sparse_output=False)
encoded = encoder.fit_transform(df[['Color']])
print(encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd
df= pd.DataFrame({
    "Size":["Small","Medium","Large","Medium","Small"]
})
print(df)
```

```python
encoder = OrdinalEncoder(
    categories=[['Small','Medium','Large']]
)
df['Size']= encoder.fit_transform(df[['Size']])
print(df)
```

```python
df= pd.DataFrame({
    "Result":["Pass","Fail","Pass","Fail","Pass"]
})
print(df)
    
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
df['Result'] = encoder.fit_transform(df['Result'])
print(df)
```

```python
decoded = encoder.inverse_transform(df['Result'])
decoded
```
