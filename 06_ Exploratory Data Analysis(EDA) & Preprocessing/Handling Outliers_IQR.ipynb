```python
#IQR is used when data distribution is skewed or data is not normal distribution 
#box plot and iqr is used
#interquartile range is a measure of the spread (dispersion )of data.
```

```python
#Q1 (first quartile) = 25tg percentile (25% of the data lies below this vlue)
#Q2 (median) = 50 th percentile
#Q3 (third quartile)= 75th percentile(75% pf the data lied below this value)
#Formula : IQR = Q3-Q1
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = [4,6,8,9,10,12,14,18,20]

Q1 = np.percentile(data,25)
Q2 = np.percentile(data,50)
Q3 = np.percentile(data,75)
```

```python
print(Q1) #25% of data are below  value 8
```

```python
print(Q2)#50% of data are below  value 10
```

```python
print(Q3)#75% of data are below  value 14
```

```python
IQR = Q3 - Q1
IQR
```

```python
import pandas as pd
df = pd.DataFrame({
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Marks":[5,50,55,60,62,65,68,70,72,150]
})
print(df)
```

```python
plt.boxplot(df['Marks'])
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Student"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
Q1 = df["Marks"].quantile(0.25)
Q2 = df["Marks"].quantile(0.50)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 -Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print(lower_bound, upper_bound)
```

```python
outlier = df[(df['Marks'] < lower_bound) | (df['Marks'] >upper_bound)]
```

```python
outlier
```

```python
#removing outlier
clean_df = df[(df['Marks'] >= lower_bound) & (df['Marks'] <=upper_bound)]
```

```python
clean_df
```

```python
import seaborn as sns
plt.figure(figsize = (6,4))
plt.boxplot(clean_df["Marks"], vert=True)

plt.title("After removing outliers")
plt.ylabel("Marks")
plt.show()
```
