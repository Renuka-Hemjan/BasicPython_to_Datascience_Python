```python
import pandas as pd

df1 = pd.DataFrame({
    "Name":["Rakesh","Sikha"],
    "Age":[25,30],
})
df2 = pd.DataFrame({
    "Name":["Anil","Priya"],
    "Age":[22,28]
})    
    
```

```python
cdf = pd.concat([df1,df2], ignore_index=True)
```

```python
cdf
```

```python
df3 = pd.DataFrame({
    "salary":[5000,7000,8000,9000]
})    
```

```python
dfc = pd.concat([cdf,df3],axis =1)
```

```python
dfc
```

```python
#merging :
employees = pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["Rakesh","Sikha","Anil"],
})
salaries = pd.DataFrame({
    "EmpID":[1,2,4],
    "Salary":[5000,7000,6000]
})
```

```python
mdf = pd.merge(employees,salaries, on = 'EmpID')
```

```python
mdf
```

```python
ldf = pd.merge(employees,salaries, on = 'EmpID', how ="left")
```

```python
ldf
```

```python
rdf = pd.merge(employees,salaries, on = 'EmpID', how ="right")
```

```python
rdf
```

```python
employees = pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["Rakesh","Sikha","Anil"],
})
salaries = pd.DataFrame({
    "EmpID":[1,2,4],
    "Salary":[5000,7000,6000]
})
```

```python
#outer merge
outer_merge = pd.merge(employees,salaries, on = 'EmpID', how ="outer")
```

```python
outer_merge
```

```python
#join it combine dataframe in basis of index
left = pd.DataFrame({
    "Name":["Rakesh","Sikha","Anil"]
}, index=[1,2,3])
right = pd.DataFrame({
    "Salary":[5000,7000,6000]
}, index=[1,2,4])
```

```python
df = left.join(right)
    
```

```python
df
```

```python
#inner join
inner = left.join(right, how="inner") #only common
print(inner)
```

```python
left.join(right, how="right")
```

```python
outer = left.join(right, how="outer") #only common
print(outer)
```
