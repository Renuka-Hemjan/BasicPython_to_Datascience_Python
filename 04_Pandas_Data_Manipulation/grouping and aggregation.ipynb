```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department":["IT","HR","IT","Finance","HR","Finance"],
    "Employee":["Rakesh","Sikha","Anil","Priya","John","Ram"],
    "Salary":[5000,6000,7000,8000,5500,9000],
    "Age":[25,30,28,35,32,40]
})    
```

```python
#sum of salary by department
sal_sum = df.groupby('Department')['Salary'].sum()
sal_sum
```

```python
print(type(sal_sum))
```

```python
sal_mean = df.groupby('Department')['Salary'].mean()
sal_mean
```

```python
dept_count = df.groupby('Department')['Salary'].count()
dept_count
```

```python
dept_min = df.groupby('Department')['Salary'].min()
dept_min
```

```python
dept_max = df.groupby('Department')['Salary'].max()
dept_max
```

```python
result = df.groupby("Department").agg({
    "Salary":["sum","mean","max"],
    "Age":["min","max"]
})   
```

```python
#calculATE diffrence of sal betn itdept and finance dept
total_it_dept_sal = df[df['Department'] == 'IT']['Salary'].sum()
total_finance_dept_sal = df[df['Department'] == 'Finance']['Salary'].sum()
difference =  total_finance_dept_sal - total_it_dept_sal 
difference
```

```python
gender = {
    "Rakesh":"Male",
    "Sikha":"Female",
    "Anil":"Male",
    "Priya":"Female",
    "John":"Male",
    "Ram":"Male"
}
df['Gender'] = df['Employee'].map(gender)
df
  
```

```python
res = df.groupby(['Department','Gender'])['Salary'].sum()
res
```

```python
result = (
    df.groupby("Department")
    .agg(
        Average_Salary=("Salary","mean"),
        Employee_Count=("Employee","count")
    )
    .reset_index()
)
print(result)
```
